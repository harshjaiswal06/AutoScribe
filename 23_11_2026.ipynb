```python
f='''
LSTMs come to the rescue to
 solve the vanishing gradient problem. It does so by ignoring (forgetting) useless data/information in the network. The LSTM will forget the data if there is no useful information from other
 inputs (prior sentence words). When new information comes, the network determines which information to be overlooked and which to be remembered.This cell state is responsible for remembering and forgetting. This is based on the context of the input. This means that some of the previous information should be remembered while some of them should be forgotten and some of the new information should be added to the memory. The first operation (X) is the pointwise operation which is nothing but multiplying the cell state by an array of [-1, 0, 1]. The information multiplied by 0 will be forgotten by the LSTM. Another operation is (+) which is responsible to add some new information to the state.

2. Forget Gate
The forget LSTM gate, as the name suggests, decides what information should be forgotten. A sigmoid layer is used to make this decision. This sigmoid layer is called the “forget gate layer”.'''
```

```python
#  we have to make tokens
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
token=Tokenizer()

token.fit_on_texts([f])
# we have to passa list because is initailly a string
```

```python
# word index will provide a dictonary in which correspondingtoevery word there is a vector
```

```python
# word index to verify that tokenizer work or not  and  every case will become lower case

token.word_index
```

```python
# the last index is answer
# count the number of unique words by counting the keys in the dictionary

len(token.word_index)
```

```python
# dividing the paragraph in lines

for sentence in f.split('\n'):
  print(sentence)
  # printing line by line printing
```

```python
#  now converting text to sequensis like lstm comes to 12 55 as the given valuer to it

for sentence in f.split('\n'):
  print(token.texts_to_sequences([sentence])[0])
  # printing line by line printing
```

```python
# next word predictor is having classification problem through lstm

# i/p             o/p
# lstm             consists
# lstm consist     of
# lstm consist of  Long
#lstm consitsoflong term memory\


#  should get converted into

# i/p                   o/p

# [12]                  [33]

# [12,33]               [34]

#  by padding we will max the value of all by taking the longest length


input_sequences=[]

for sentence in f.split('\n'):
  tokened_sentence=token.texts_to_sequences([sentence])[0]
  for i in range(1,len(tokened_sentence)):
    input_sequences.append(tokened_sentence[:i+1])

# Let's print the generated sequences to verify
print(f"Total input sequences: {len(input_sequences)}")
print("First 5 sequences:", input_sequences[:5])
```

```python
x_len=[]
```

```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

# We pad the sequences to have the same length
# Let's find the maximum length in our sequences dynamically
max_len = max([len(x) for x in input_sequences])

padded_input_sequences = pad_sequences(input_sequences, maxlen=max_len, padding='pre')
padded_input_sequences
```

```python
# Now split into features (X) and labels (y)
# X contains all tokens except the last one of each padded sequence
# y contains the last token of each padded sequence
X = padded_input_sequences[:, :-1]
y = padded_input_sequences[:, -1]
```

```python
# Verify the shape of X
print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python
#  this is classfication problem
#  soit is a multiclass classification we have to use softmax functipn and doin one hot encodeing make the only one the top

from keras.utils import to_categorical
y=to_categorical(y,num_classes=len(token.word_index)+1)
y.shape
#
```

```python
#  now our data is ready
```

```python
from keras.layers import Dense,Input,LSTM,Embedding
from keras.models import Sequential
```

```python
model=Sequential();

model.add(Input(shape=(28,)))
# our x means this 28 is the input dimension

model.add(Embedding(95,100))
# 94 is the total unique number

#  through embedding we are converting that word into vector and what os the advantage so it help finding the relation in words for ex there is a word bat where bat is playing bat and a animal bat so if we use one hot encoding or integer encoding si it will give the same number to bith the bat but in embedding will recognize the bat with its nearest sentenece

model.add(LSTM(100))
# vector size = hidden layer

model.add(Dense(95,activation='softmax'))
# howmany layers so equal to the number of words
model.summary()
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100)
```
